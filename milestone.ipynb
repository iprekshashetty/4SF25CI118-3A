```python
import matplotlib.pyplot  as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=data ,x='Vehicle',y='Delivery_time')
plt.title("Vehicle v/s Delivery_time")
plt.xlabel("Delivery Time")
plt.ylabel("Vehicke Used");
plt.show()

```

```python
plt.figure(figsize=(8,4))
data.groupby('Area')['Order_Category'].size().plot(kind='line',marker='o',color='Blue')
plt.title("Orders in  particular Area")
plt.xlabel('Area')
plt.ylabel('Order Category')
plt.grid(True)
plt.show()

```

```python
plt.figure(figsize=(8,10))
data.groupby('Order_Category')['month'].size().plot(kind='hist',color='green')
plt.title("Orders in particular month")
plt.xlabel('Order Category')
plt.ylabel('Month')
plt.show()
```

```python
data['Vehicle'].value_counts().plot(kind='pie',color=['teal','blue','orange'])
plt.title("Vehicles")
plt.show()
```

```python
daily_orders=data.groupby('Order_Date').size()
plt.figure(figsize=(10,5))
plt.plot(daily_orders.index,daily_orders.values,color='blue',linewidth=2)
```
